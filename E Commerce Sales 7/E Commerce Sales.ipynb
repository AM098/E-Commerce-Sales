```python
import pandas as pd 
import plotly.express as pxi
import plotly.io as pio
import plotly.colors as colors
import plotly.graph_objects as go 

pio.templates.default = "plotly_white"  #pio.templates.default = "plotly_white": Default theme white rakha gaya hai graphs ke liye
pio.renderers.default = "browser"
```

```python
data = pd.read_csv("Superstore.csv", encoding='latin-1') 
# encoding='latin-1': used to read special characters in the file correctly.
data.head()
```

```python
data.describe()
```

```python
'''
The dataset has an order date column. We can use this column
to create new columns like order month, order year, and order day,
which will be very valuable for sales and profit analysis
according to time periods. So let's add these columns:
'''
```

```python
data.head()
```

```python
data.info()
```

```python
# Converting Date Columns
```

```python
# Date Conversion: converts the Order Date and Ship Date columns to datetime format for date-based analysis.
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date']) 

```

```python
# Adding New Date-Based Columns
```

```python
# Order Month: extracts the month from the order date.
# Order Year: extracts the year from the order date.
# Order Day of Week: extracts the day of the week (0 for Monday, 6 for Sunday).

data['Order Month'] = data['Order Date'].dt.month 
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data.head()
```

```python
# Monthly Sales Analysis
```

```python
# Data Grouping:
# data.groupby('Order Month')['Sales'].sum() calculates the total sales for each month.
# .reset_index() converts the result back into a regular DataFrame (a structured table).
# px.line: creates a line chart to show the monthly sales trend.


sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
fig = px.line(sales_by_month, 
              x='Order Month', 
              y='Sales', 
              title='Monthly Sales Analysis')
fig.show()

# fig.show(): displays the graph.
```

```python

                                                          #   Sales Analysis by Category
```

```python
# groupby('Category'): calculates the total sales for each category.
# Pie Chart:
# px.pie: shows the sales proportions as a pie chart.
# hole=0.5: creates a donut-style chart.
# Pastel Colors: uses a soft color palette in the chart.

sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()


fig = px.pie(sales_by_category, 
             values='Sales', 
             names='Category', 
             hole=0.5, 
             color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text='Sales Analysis by Category', title_font=dict(size=24))

fig.show()
```

```python
#  OUTPUT == fig.show(): displays the browser.
```

```python
                                         # Sales Analysis by Sub-Category
```

```python
    # Monthly Profit Analysis
```

```python
profit_by_month = data.groupby('Order Month')['Profit'].sum().reset_index()
fig = px.line(profit_by_month, 
              x='Order Month', 
              y='Profit', 
              title='Monthly Profit Analysis')
fig.show()
```

```python
#  OUTPUT == fig.show(): displays the browser.
```

```python
    # Profit Analysis by Category
```

```python
profit_by_category = data.groupby('Category')['Profit'].sum().reset_index()

fig = px.pie(profit_by_category, 
             values='Profit', 
             names='Category', 
             hole=0.5, 
             color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text='Profit Analysis by Category', title_font=dict(size=24))

fig.show()
```

```python
#  OUTPUT == fig.show(): displays the browser.
```

```python
     # Profit Analysis by Sub-Category
```

```python
profit_by_subcategory = data.groupby('Sub-Category')['Profit'].sum().reset_index()
fig = px.bar(profit_by_subcategory, x='Sub-Category', 
             y='Profit', 
             title='Profit Analysis by Sub-Category')
fig.show()
```

```python
#  OUTPUT == fig.show(): displays the browser.
```

```python
      # Sales and Profit Analysis by Customer Segment
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()

color_palette = colors.qualitative.Pastel

fig = go.Figure()
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'], 
                     y=sales_profit_by_segment['Sales'], 
                     name='Sales',
                     marker_color=color_palette[0]))

fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'], 
                     y=sales_profit_by_segment['Profit'], 
                     name='Profit',
                     marker_color=color_palette[1]))

fig.update_layout(title='Sales and Profit Analysis by Customer Segment',
                  xaxis_title='Customer Segment', yaxis_title='Amount')

fig.show()
```

```python
#  OUTPUT == fig.show(): displays the browser.
```

```python
     #  analyse sales-to-profit ratio
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```
